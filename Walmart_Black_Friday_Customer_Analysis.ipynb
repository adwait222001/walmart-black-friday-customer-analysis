How do customer demographics and spending patterns vary across gender, age groups, and marital status?

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('walmart_data.csv')

# Basic gender count
gender_count = data.groupby('Gender')['User_ID'].count()
print(gender_count)

# Separate male and female data
female = data[data['Gender'] == 'F']
male = data[data['Gender'] == 'M']

# Number of users by age for each gender
female_count = female.groupby('Age')['User_ID'].count()
male_count = male.groupby('Age')['User_ID'].count()

# Number of married/unmarried users by age for each gender
female_marriage_status = female.groupby(
    ['Marital_Status', 'Age']
)['User_ID'].count()

male_marriage_status = male.groupby(
    ['Marital_Status', 'Age']
)['User_ID'].count()

print("The female marriage status:\n", female_marriage_status)
print("The male marriage status:\n", male_marriage_status)

# Spending by marital status for females
spending_per_marital_status_per_female = \
    female.groupby('Marital_Status')['Purchase'].sum()

spending_per_marital_status_per_female.index = \
    spending_per_marital_status_per_female.index.map({
        0: 'Unmarried',
        1: 'Married'
    })

# Spending by marital status for males
spending_per_marital_status_per_male = \
    male.groupby('Marital_Status')['Purchase'].sum()

spending_per_marital_status_per_male.index = \
    spending_per_marital_status_per_male.index.map({
        0: 'Unmarried',
        1: 'Married'
    })

print(
    "Spending as per marital status for female:\n",
    round(spending_per_marital_status_per_female / 100000, 3)
)

print(
    "Spending as per marital status for male:\n",
    round(spending_per_marital_status_per_male / 100000, 3)
)

# Male vs female users by age
plt.plot(
    male_count.index,
    male_count.values,
    color='black',
    label='Male'
)

plt.plot(
    female_count.index,
    female_count.values,
    color='red',
    label='Female'
)

plt.xlabel('Age-group')
plt.ylabel('User count')
plt.title('(Age-group)-(User-count)')
plt.legend()
plt.show()

# Married/unmarried users by age for females and males
plt.plot(
    female_marriage_status.xs(0, level='Marital_Status'),
    color='red',
    label='Female - Unmarried'
)

plt.plot(
    female_marriage_status.xs(1, level='Marital_Status'),
    color='purple',
    label='Female - Married'
)

plt.plot(
    male_marriage_status.xs(0, level='Marital_Status'),
    color='black',
    label='Male - Unmarried'
)

plt.plot(
    male_marriage_status.xs(1, level='Marital_Status'),
    color='brown',
    label='Male - Married'
)

plt.xlabel('Age')
plt.ylabel('User Count')
plt.title('User Count by Age, Gender and Marital Status')
plt.legend()
plt.show()
```

Does customer purchase amount significantly differ across occupation categories and years of stay in the current city, and is there an interaction effect between occupation and years of stay?

Hypotheses

1. Occupation Category

H₀: There is no significant difference in purchase amount among different occupation categories.
H₁: There is a significant difference in purchase amount among different occupation categories.

2. Stay in Current City

H₀: There is no significant difference in purchase amount among customers based on their years of stay in the current city.
H₁: There is a significant difference in purchase amount among customers based on their years of stay in the current city.

3. Interaction Effect

H₀: There is no significant interaction effect between occupation category and years of stay in the current city on purchase amount.
H₁: There is a significant interaction effect between occupation category and years of stay in the current city on purchase amount.

```python
import pandas as pd
from statsmodels.formula.api import ols
import statsmodels.api as sm
data = pd.read_csv('walmart_data.csv')
sub_chart = data[['Occupation_category','Stay_In_Current_City_Years','Purchase']]
#H0 for occupation_category:there is no significant difference in purchase of products between the occupation_category
#H1 for occupation_category:there is signifcant difference in purchase of products between the occupation_category

#H0 for stay_in_current_city_years:there is no signifcant difference in purchase of products due to staying in a city
#H1 for stay_in_current_city_years:there is signinfcant differnece in purchase of products due to staying in a city

#together effect
#H0: there is no interraction effect between the occupation_category and stay_in_current_city_years on the purchase of the product
#H1: there is signicant interraction effect between the occupation_category and stay_in_current_city_years
test = ols('Purchase ~ C(Occupation_category) * C(Stay_In_Current_City_Years)', data=sub_chart).fit()
annova_table = sm.stats.anova_lm(test, typ=2)
print(annova_table)
```

Does customer purchase amount significantly differ based on gender and product category, and is there a significant interaction effect between gender and product category on purchase amount?

1. Gender

H₀ (Null Hypothesis):
There is no significant difference in purchase amount between different genders.

H₁ (Alternative Hypothesis):
There is a significant difference in purchase amount between different genders.

2. Product Category

H₀ (Null Hypothesis):
There is no significant difference in purchase amount among different product categories.

H₁ (Alternative Hypothesis):
There is a significant difference in purchase amount among different product categories.

3. Interaction Effect — Gender × Product Category

H₀ (Null Hypothesis):
There is no significant interaction effect between gender and product category on purchase amount.

H₁ (Alternative Hypothesis):
There is a significant interaction effect between gender and product category on purchase amount.

```python
import pandas as pd
from statsmodels.formula.api import ols
import statsmodels.api as sm
data = pd.read_csv('walmart_data.csv')
sub_chart = data[['Gender','Product_Category','Purchase']]
test = ols('Purchase ~ C(Gender) * C(Product_Category)', data=sub_chart).fit()
annova_table = sm.stats.anova_lm(test, typ=2)
print(annova_table)
annova_table
```

Using the Central Limit Theorem, analyze and compare the distribution of sample mean purchase amounts for male and female customers.”

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

data = pd.read_csv('walmart_data.csv')

male = data[data['Gender'] == 'M']['Purchase']
female = data[data['Gender'] == 'F']['Purchase']
#means of condition
male_mean = male.mean()
female_mean = female.mean()

#empty array for storage
male_sample_means = []
female_sample_means = []

#std for female and male
male_std = male.std()
female_std = female.std()

for i in range(1000):

    male_sample = male.sample(100, replace=True)

    male_sample_means.append(male_sample.mean())

    female_sample = female.sample(100, replace=True)

    female_sample_means.append(female_sample.mean())


# Confidence Interval
n = 100
z = 1.96

# Male Confidence Interval
male_sample_mean = np.mean(male_sample_means)
male_margin_of_error = z * (male_std / np.sqrt(n))
male_lower = male_sample_mean - male_margin_of_error
male_upper = male_sample_mean + male_margin_of_error

# Female Confidence Interval
female_sample_mean = np.mean(female_sample_means)
female_margin_of_error = z * (female_std / np.sqrt(n))
female_lower = female_sample_mean - female_margin_of_error
female_upper = female_sample_mean + female_margin_of_error

#printing the output
print("Male Sample Mean:", male_sample_mean)
print("Male 95% Confidence Interval:", male_lower, male_upper)
print("Female Sample Mean:", female_sample_mean)
print("Female 95% Confidence Interval:", female_lower, female_upper)
#plotting the graph
sns.histplot(male_sample_means, kde=True, label='Male')
sns.histplot(female_sample_means, kde=True, label='Female')
plt.title("Distribution of Sample Means of Purchase by Gender")
plt.xlabel("Sample Mean Purchase")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

Using the Central Limit Theorem, how does the average purchase amount differ across different age groups based on their sampling distributions and 95% confidence intervals?

```python
import pandas as pd
import numpy as np

data = pd.read_csv('walmart_data.csv')




male = data[data['Gender'] == 'M']['Purchase']
female = data[data['Gender'] == 'F']['Purchase']

male_sample_means = []
female_sample_means = []

male_std = male.std()
female_std = female.std()

for i in range(1000):

    male_sample = male.sample(100, replace=True)
    male_sample_means.append(male_sample.mean())

    female_sample = female.sample(100, replace=True)
    female_sample_means.append(female_sample.mean())




data['Stay_In_Current_City_Years'] = data[
    'Stay_In_Current_City_Years'
].replace('4+', '4').astype(int)

male_city = data[data['Gender'] == 'M']['Stay_In_Current_City_Years']
female_city = data[data['Gender'] == 'F']['Stay_In_Current_City_Years']

male_city_sample_means = []
female_city_sample_means = []

male_city_std = male_city.std()
female_city_std = female_city.std()

for i in range(1000):

    male_city_sample = male_city.sample(100, replace=True)
    male_city_sample_means.append(male_city_sample.mean())

    female_city_sample = female_city.sample(100, replace=True)
    female_city_sample_means.append(female_city_sample.mean())



married = data[data['Marital_Status'] == 1]['Purchase']
single = data[data['Marital_Status'] == 0]['Purchase']

married_sample_means = []
single_sample_means = []

married_std = married.std()
single_std = single.std()

for i in range(1000):

    married_sample = married.sample(100, replace=True)
    married_sample_means.append(married_sample.mean())

    single_sample = single.sample(100, replace=True)
    single_sample_means.append(single_sample.mean())




data['Age_Group'] = data['Age'].replace({
    '0-17': '0-17',
    '18-25': '18-25',
    '26-35': '26-35',
    '36-45': '36-50',
    '46-50': '36-50',
    '51-55': '51+',
    '55+': '51+'
})

age_0_17 = data[data['Age_Group'] == '0-17']['Purchase']
age_18_25 = data[data['Age_Group'] == '18-25']['Purchase']
age_26_35 = data[data['Age_Group'] == '26-35']['Purchase']
age_36_50 = data[data['Age_Group'] == '36-50']['Purchase']
age_51_plus = data[data['Age_Group'] == '51+']['Purchase']

age_0_17_sample_means = []
age_18_25_sample_means = []
age_26_35_sample_means = []
age_36_50_sample_means = []
age_51_plus_sample_means = []

age_0_17_std = age_0_17.std()
age_18_25_std = age_18_25.std()
age_26_35_std = age_26_35.std()
age_36_50_std = age_36_50.std()
age_51_plus_std = age_51_plus.std()

for i in range(1000):

    age_0_17_sample = age_0_17.sample(100, replace=True)
    age_0_17_sample_means.append(age_0_17_sample.mean())

    age_18_25_sample = age_18_25.sample(100, replace=True)
    age_18_25_sample_means.append(age_18_25_sample.mean())

    age_26_35_sample = age_26_35.sample(100, replace=True)
    age_26_35_sample_means.append(age_26_35_sample.mean())

    age_36_50_sample = age_36_50.sample(100, replace=True)
    age_36_50_sample_means.append(age_36_50_sample.mean())

    age_51_plus_sample = age_51_plus.sample(100, replace=True)
    age_51_plus_sample_means.append(age_51_plus_sample.mean())

groups = {

    # Gender - Purchase
    'Male Purchase': (male_sample_means, male_std),
    'Female Purchase': (female_sample_means, female_std),

    # Gender - Stay in Current City
    'Male Stay in City': (male_city_sample_means, male_city_std),
    'Female Stay in City': (female_city_sample_means, female_city_std),

    # Marital Status - Purchase
    'Married Purchase': (married_sample_means, married_std),
    'Single Purchase': (single_sample_means, single_std),

    # Age - Purchase
    'Age 0-17 Purchase': (age_0_17_sample_means, age_0_17_std),
    'Age 18-25 Purchase': (age_18_25_sample_means, age_18_25_std),
    'Age 26-35 Purchase': (age_26_35_sample_means, age_26_35_std),
    'Age 36-50 Purchase': (age_36_50_sample_means, age_36_50_std),
    'Age 51+ Purchase': (age_51_plus_sample_means, age_51_plus_std)
}


confidence_levels = {
    90: 1.645,
    95: 1.96,
    99: 2.576
}

n = 100

confidence_intervals = []

for group, (sample_means, std) in groups.items():

    sample_mean = np.mean(sample_means)

    for confidence, z in confidence_levels.items():

        margin = z * (std / np.sqrt(n))

        lower = sample_mean - margin
        upper = sample_mean + margin

        confidence_intervals.append([
            group,
            confidence,
            sample_mean,
            lower,
            upper
        ])




ci_table = pd.DataFrame(
    confidence_intervals,
    columns=[
        'CLT Group',
        'Confidence Level',
        'Sample Mean',
        'Lower Limit',
        'Upper Limit'
    ]
)

# Round values for a cleaner table
ci_table['Sample Mean'] = ci_table['Sample Mean'].round(2)
ci_table['Lower Limit'] = ci_table['Lower Limit'].round(2)
ci_table['Upper Limit'] = ci_table['Upper Limit'].round(2)

ci_table['Confidence Level'] = (
    ci_table['Confidence Level'].astype(str) + '%'
)

print("\nALL CLT CONFIDENCE INTERVALS")

display(ci_table)


# =========================================================
# 8. 95% CONFIDENCE INTERVALS ONLY
# =========================================================

ci_95 = ci_table[
    ci_table['Confidence Level'] == '95%'
].reset_index(drop=True)

print("\n95% CONFIDENCE INTERVALS")

display(ci_95)
```

How do the mean and median purchase amounts compare, and what does the outlier analysis reveal about the distribution of customer purchase amounts in the Walmart dataset?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = pd.read_csv('walmart_data.csv')
print("Statistical Summary of Purchase Amount:")
print(data['Purchase'].describe())
mean_purchase = data['Purchase'].mean()
median_purchase = data['Purchase'].median()
print("\nMean Purchase Amount:", mean_purchase)
print("Median Purchase Amount:", median_purchase)
print("\nMean vs Median:")
if mean_purchase > median_purchase:
    print("The mean is higher than the median, indicating that higher purchase values may be influencing the average.")
elif mean_purchase < median_purchase:
    print("The mean is lower than the median, indicating that lower purchase values may be influencing the average.")
else:
    print("The mean and median are equal.")
Q1 = data['Purchase'].quantile(0.25)
Q3 = data['Purchase'].quantile(0.75)
IQR = Q3 - Q1
print("\nFirst Quartile (Q1):", Q1)
print("Third Quartile (Q3):", Q3)
print("Interquartile Range (IQR):", IQR)
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
print("\nLower Outlier Limit:", lower_limit)
print("Upper Outlier Limit:", upper_limit)
outliers = data[
    (data['Purchase'] < lower_limit) |
    (data['Purchase'] > upper_limit)
]
print("\nNumber of Outliers:", len(outliers))
print("\nOutlier Purchase Values:")
print(outliers['Purchase'].head(10))
plt.figure(figsize=(8, 5))
plt.boxplot(data['Purchase'])
plt.title('Boxplot of Purchase Amount')
plt.ylabel('Purchase Amount')
plt.show()
```
